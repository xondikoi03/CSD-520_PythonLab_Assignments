## Problem 1:

Create a tuple containing the marks of a student in five subjects:

Write a Python program to:
1. Display all marks.
2. Calculate total marks.
3. Calculate percentage.
4. Find the highest and lowest marks.
5. Count the number of subjects in which the student scored more than 75.
6. Display the marks in sorted order.

````python
## Create a tuple of marks of a student in five subjects:
marks = (85, 90, 80, 92, 88)

## User defined function:
def display_Student_Details(marks):

    # 2. Calculate total marks:
    total_marks = sum(marks)
    # 3. Calculate average marks ane percentage:
    average_marks = total_marks / len(marks)
    percentage = (total_marks / 500) * 100
    # 4. Display the minimum and maximum marks:
    min_marks = min(marks)
    max_marks = max(marks)
    # 5. Number of subjects above 75:
    mark_counter = 0
    for mark in marks:
        if mark > 75:
            mark_counter += 1

    # 6. Display the marks in sorted order:
    sorted_marks = sorted(marks)    

    print("Marks in five subjects:", marks)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Percentage:", percentage)
    print("Minimum Marks:", min_marks)
    print("Maximum Marks:", max_marks)
    print("Number of subjects above 75:", mark_counter)
    print("Marks in sorted order:", sorted_marks)
````

````python
## Call the function:
display_Student_Details(marks)
````

````python
marks2 = (70, 60, 71, 65, 20)
display_Student_Details(marks2)
````

## Problem 2:

Two groups of students have learned different programming technologies.

Create two sets:
```
Group A = {Python, Java, SQL, Pandas, NumPy}
Group B = {Python, SQL, R, Tableau, Pandas}
```
Write a Python program to find:

1. Skills common to both groups.
2. Skills learned only by Group A.
3. Skills learned only by Group B.
4. All unique skills learned by both groups.
5. Whether both groups have any common skills.

````python
## Create two sets:
Group_A = {"Python", "Java", "SQL", "Pandas", "Numpy"}
Group_B = {"Python", "SQL", "R", "Tableau", "Pandas"}

# 1. Skills common to both groups:
common_skills = Group_A.intersection(Group_B)
print(f"Skills common to both groups: {common_skills}")

# 2. Skills learned only by group A:
skills_only_group_A = Group_A.difference(Group_B)
print(f"Skills learned only by group A: {skills_only_group_A}")

# 3. Skills learned only by group B:
skills_only_group_B = Group_B.difference(Group_A)
print(f"Skills learned only by group B: {skills_only_group_B}")

# 4. All unique skills learned by both groups:
all_unique_skills = Group_A.union(Group_B)
print(f"All unique skills learned by both groups: {all_unique_skills}")

# 5. Whether both group have any skills in common or not:
if common_skills:
    print("Both groups have skills in common.")
else:
    print("Both groups do not have any skills in common.")
````

## Problem 3:

Create a dictionary to maintain the inventory of products in a store.

Example:
```
inventory = {
    "Laptop": 15,
    "Mouse": 40,
    "Keyboard": 25,
    "Monitor": 10
}
```
Write a menu-driven Python program to:

1. Add a new product.
2. Update the quantity of an existing product.
3. Remove a product.
4. Display all products.
5. Search for a product.
6. Display products whose quantity is less than 15.


````python
# Create a dictionary for inventory of products in a store:
inventory = {
    "Laptop": 15,
    "Mouse": 40,
    "Keyboard": 25,
    "Monitor": 10,
    "Printer": 5
}

# User defined functions:
def inventoru_info(inventory_dict):
    # Menu for user to choose an option:
    # 1. Add a new product to the inventory.
    # 2. Update the quantity of an existing product.
    # 3. Delete a product from the inventory.
    # 4. Display the inventory.
    # 5. Search for a product in the inventory.
    # 6. Dis[play products under a certain quantity.
    # 7. Exit the program.

    while True:
        print("\nInventory Management System")
        print("1. Add a new product to the inventory.")
        print("2. Update the quantity of an existing product.")
        print("3. Delete a product from the inventory.")
        print("4. Display the inventory.")
        print("5. Search for a product in the inventory.")
        print("6. Display products under a certain quantity.")
        print("7. Exit the program.")

        choice = input("Enter your choice (1-7): ")

        if choice == '1':
            product_name = input("Enter the name of the product: ")
            quantity = int(input("Enter the quantity of the product: "))
            inventory_dict[product_name] = quantity
            print(f"{product_name} added to inventory with quantity {quantity}.")
        elif choice == '2':
            product_name = input("Enter the name of the product: ")
            if product_name in inventory_dict:
                quantity = int(input("Enter the new quantity of the product: "))
                inventory_dict[product_name] = quantity
                print(f"{product_name} updated with quantity {quantity}.")
            else:
                print(f"{product_name} not found in inventory.")
        elif choice == '3':
            product_name = input("Enter the name of the product to delete: ")
            if product_name in inventory_dict:
                del inventory_dict[product_name]
                print(f"{product_name} deleted from inventory.")
            else:
                print(f"{product_name} not found in inventory.")
        elif choice == '4':
            print("Current Inventory:")
            for product, quantity in inventory_dict.items():
                print(f"{product}: {quantity}")
        elif choice == '5':
            product_name = input("Enter the name of the product to search for: ")
            if product_name in inventory_dict:
                print(f"{product_name}: {inventory_dict[product_name]}")
            else:
                print(f"{product_name} not found in inventory.")
        elif choice == '6':
            threshold = int(input("Enter the minimum quantity to display: "))
            print("Products with quantity below", threshold, ":")
            for product, quantity in inventory_dict.items():
                if quantity < threshold:
                    print(f"{product}: {quantity}")
        elif choice == '7':
            print("Exiting the program.")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 7.")
````
