## Problem 1:

Create a menu-driven calculator program using the `elif` ladder: 

1. Addition
2. Subtraction
3. Multiplication
4. Division

```python
while True:
    print("------------- ARITHMETIC CALCULATOR -------------")
    print("Select operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("Enter 'q/Q' to quit the program.")

    opr = input("Enter operation (1/2/3/4/q): ")
    if opr in ['q', 'Q']:
        print("Thank you for using the arithmetic calculator!")
        break

    if opr in ['1', '2', '3', '4']:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        if opr == '1':
            print("Addition selected.")
            result = num1 + num2
            print(f"The result of addition is: {result}")
        elif opr == '2':
            print("Subtraction selected.")
            result = num1 - num2
            print(f"The result of subtraction is: {result}")
        elif opr == '3':
            print("Multiplication selected.")
            result = num1 * num2
            print(f"The result of multiplication is: {result}")
        elif opr == '4':
            print("Division selected.")
            if num2 != 0:
                result = num1 / num2
                print(f"The result of division is: {result}")
            else:
                print("Error: Division by zero is not allowed.")
    else:
        print("Invalid operation. Please select a valid operation (1/2/3/4/q).")
```

## Problem 2: 
Accept weight (in kgs) and height (in m), calculate BMI and classify:
1. BMI < 18.5 $\rightarrow$ Underweight
2. 18.5 $\leq$ BMI $<$ 24.9 $\rightarrow$ Normal
3. 25 $\leq$ BMI $<$ 29.9 $\rightarrow$ Overweight 
4. BMI $\geq$ 30 $\rightarrow$ Obese 

```python
while True:
    print("------------- BMI CALCULATOR -------------")
    weight = float(input("Enter your weight in kilograms: "))
    height = float(input("Enter your height in meters: "))

    if height <= 0:
        print("Error: Height must be greater than zero.")
        continue

    bmi = weight / (height ** 2)
    print(f"Your BMI is: {bmi:.2f}")

    if bmi < 18.5:
        print("BMI indicates underweight.")
    elif 18.5 <= bmi < 24.9:
        print("BMI indicates a normal weight.")
    elif 25 <= bmi < 29.9:
        print("BMI indicates overweight.")
    else:
        print("BMI indicates obesity.")

    cont = input("Do you want to calculate BMI again? (y/n): ")
    if cont.lower() != 'y':
        print("Thank you for using the BMI calculator!")
        break
```

## Problem 3: 

Accept number of units consumed and calculate the total bill using the following:

1. Up to 100 units: ₹5/unit
2. 101-200 units: ₹6/unit
3. 201-300 units: ₹8/unit
4. Above 300 units: ₹10/unit

```python
while True:
    print("--------------BILL CALCULATOR-----------------")
    print("Enter 'q' to quit")
    
    units = input("Enter the number of units consumed (q to quit): ")
    if units.lower() != 'q':
        print("Units Consumed:", units)
    if units in ['q', 'Q']:
        print("Thank you for using the bill calculator! Exiting...")
        break

    # TRY STATEMENT FOR ERROR HANDLING:
    try:
        units = float(units)

        # Check for negative units to avoid invalid input:
        if units < 0:
            print("Units consumed cannot be negative.")
            continue # Starts the while loop again for new input

        # BILL CALCULATION:
        if units < 0:
            print("Units consumed cannot be negative.")
            continue
        elif units <= 100:
            bill_amount = units * 5
        elif units <= 200:
            bill_amount = (100 * 5) + ((units - 100) * 6)
        elif units <= 300:
            bill_amount = (100 * 5) + (100 * 6) + ((units - 200) * 8)
        else:  # above 300 units
            bill_amount = (100 * 5) + (100 * 6) + (100 * 8) + ((units - 300) * 10)

        print(f"Your total bill amount is: {bill_amount:.2f} INR")

    # IF USER ENTERS NON-NUMERIC VALUES FOR UNITS, THIS WILL CATCH THE ERROR:        
    except ValueError:
        print("Invalid input. Please enter a numeric value for units consumed.")
```

## Problem 5: 

Write a program to accept a range from the user and display following:
1. All prime numbers within the range
2. Sum of squares of all numbers within the range

```python
# 1. All prime numbers with a range given by the user:
n = int(input("Enter the upper limit to find all prime numbers: "))
print(f"Prime numbers up to {n}:")

# For loop that runs from 2 to n (inclusive) to check for prime numbers:
for num in range(2, n + 1):
    # Setting a flag to check if the number is prime:
    # is_prime: True -> prime
    # is_prime: False -> not prime
    is_prime = True

    # Check if num is divisible by any number from 2 to the square root of num: (after the square root, factors repeat) 
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0: # If num is divisible by i, then it is not prime:
            is_prime = False
            break
    if is_prime: 
        print(f'Prime number: {num}')
```

```python
# 2. Sum of the squares of the first numbers within a range given by the user:
n = int(input("Enter the upper limit to calculate the sum of squares: "))
# Initialize sum_of_squares to 0:
sum_of_squares = 0
for i in range(1, n + 1):
    sum_of_squares += i ** 2 # Adding the square of i to sum_of_squares in each iteration
print(f"The sum of squares from 1 to {n} is: {sum_of_squares}")
```

## Problem 7:

Write a program to accept a number and display:
1. Reverse of a number
2. Check the number is palindrome or not?

```python
# 1. Accept a number:
number = input("Enter a number: ")
# Print the reverse of the number:
print(f"Reverse of {number}: {number[::-1]}")
```

```python
# 2. Check if number is a palindrome or not:
# from pydoc import text
number = input("Enter a number: ") 
# Check if the number is equal to its reverse:
if number == number[::-1]:
    print(f"{number} is a palindrome.")
else:
    print(f"{number} is not a palindrome.")
```

## Problem 8:

Write a program to accept a string and display count of vowels and consonants in a string.

```python
# Program to accept a string and count the vowels:
str = input("Enter a string: ")
vowels = "aeiouAEIOU"
vowel_counter = 0
consonant_counter = 0
for char in str:
    if char in vowels:
        vowel_counter += 1
    else:
        consonant_counter += 1
print(f"The user entered string is: {str}")
print(f"The number of vowels in the string is: {vowel_counter}")
print(f"The number of consonants in the string is: {consonant_counter}")
```
