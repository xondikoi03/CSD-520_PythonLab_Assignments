# Assignment 9: Data Visualisation

## Problem 1: Bar Chart - Category Wise Sales
Download the given e-commerce sales dataset and create a bar chart to visualize the total sales/amount for each product category. Add a suitable title and labels for the X-axis and Y-axis. 

## Problem 2: Pie Chart – Regional Sales Distribution

Using the same dataset, create a pie chart showing the percentage distribution of total sales across different regions. Display suitable labels and percentages. 

## Problem 3: Histogram – Distribution of Sales Amount

## Problem 4: Word Cloud – Product Names
