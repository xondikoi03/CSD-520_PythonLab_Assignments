## Problem 2: 

Write a program: 

- Create a class `employee` having `empid` and `name`.
- Create a class `salary` having the members `basic_salary`, `TA`, `HRA`.
- Create a class `TotalSal` having the function `Gross_Sal()` to calculate the gross salray of an employee.

Define the appropriate constructors.

**Note:**
$$
\text{Gross Salary = Basic Salary + TA + HRA}
$$

```python
# Employee Class:
class Employee:
    def __init__(self, emp_id, emp_name):
        self.emp_id = emp_id
        self.emp_name = emp_name

# Salary Class:
class Salary(Employee): # Inheriting from Employee class
    def __init__(self, emp_id, emp_name, basic_salary, ta, hra):
        super().__init__(emp_id, emp_name) # Calling the constructor of the parent class
        self.basic_salary = basic_salary
        self.ta = ta
        self.hra = hra

# TotalSalary Class:
class TotalSal(Salary): # Inheriting from Salary class
    def __init__(self, emp_id, emp_name, basic_salary, ta, hra):
        super().__init__(emp_id, emp_name, basic_salary, ta, hra) 
    # Method to calculate and display gross salary
    def Gross_Sal(self, gross_salary=None): 
        gross_salary = self.basic_salary + self.ta + self.hra 
        print(f"--- Employee Details ---")
        print(f"Employee ID: {self.emp_id}") 
        print(f"Employee Name: {self.emp_name}") 
        print(f"Basic Salary: {self.basic_salary}") 
        print(f"TA: {self.ta}") 
        print(f"HRA: {self.hra}") 
        print(f"Gross Salary: {gross_salary}") 
```

### Displaying Employee Details:

```python
# Employee 1:
employee_1 = TotalSal(
    emp_id = 101,
    emp_name = "Raj Handique",
    basic_salary = 50000,
    ta = 5000,
    hra = 10000
)
# Calling the method to display employee details and gross salary
employee_1.Gross_Sal() 

# Employee 2:
employee_2 = TotalSal(
    emp_id = 102,
    emp_name = "Ananya Das",
    basic_salary = 60000,
    ta = 6000,
    hra = 12000
)
employee_2.Gross_Sal() 
```

## Problem 3:

Create a class `BankAccount` with:

- Account Number
- Account Holder Name
- Balance

Implement methods:

- `deposit()`
- `withdraw()`
- `display_balance()`

**Ensure that a customer cannot withdraw more money than the available balance.**

Create an object and perform multiple deposit and withdrawal operations.

```python
# Bank Account Class:

class Bank_Account():
    def __init__(self, acc_number, acc_holder_name, acc_balance=0.0):
        self.acc_number = acc_number
        self.acc_holder_name = acc_holder_name
        self.acc_balance = float(acc_balance)

    # User Defined Methods:

    # 1. Deposit Method:
    def deposit(self, amount):
        if amount > 0:
            self.acc_balance += amount
            print(f"Successfully deposited {amount} into account {self.acc_number}. New balance: {self.acc_balance}")
        else:
            print("Invalid deposit amount.")

    # 2. Withdraw Method:
    def withdraw(self, amount):
        if amount <= 0:
            print(f"Invalid withdrawal amount.")
        elif amount > self.acc_balance:
            print(
                f"Insufficient funds in account {self.acc_number}! Cannot withdraw {amount}. Current balance: {self.acc_balance}"
            )
        else:
            self.acc_balance -= amount
            print(f"Successfully withdrew {amount} from account {self.acc_number}. New balance: {self.acc_balance}")

    # 3. Display Method:
    def display_balance(self):
        print(f"\n --- Account Information: ---")
        print(f"Account Number: {self.acc_number}")
        print(f"Account Holder: {self.acc_holder_name}")
        print(f"Current Balance: {self.acc_balance}")
```

### Bank Account Details:

```python
# Bank Account 1:
account_1 = Bank_Account(
    acc_number = "CB0567809",
    acc_holder_name = "Raj Handique",
    acc_balance = 4678.25
)

# Bank Account 2:
account_2 = Bank_Account(
    acc_number = "CB0567810",
    acc_holder_name = "Ananya Das",
    acc_balance = 10274.12
)
```

```python
# Depositing money into account 1:
account_1.deposit(1500.00) 
account_1.display_balance() 
```

## Problem 4:

Create a base class `Shape` with a method `display()`. Also create two derived classes:

- Circle
- Rectangle

Implement suitable methods to calculate the area of a shape. Create obects of both classes and display their areas.

```python
import math

# Base Class - Shape:
class Shape:
    def display(self):
        print("This is a shape.")

# Derived Class - Circle:
class Circle(Shape):
    def __init__(self, radius):
        self.radius = float(radius)

    def area(self):
        return math.pi * self.radius ** 2

    def display(self):
        print(f"Shape: Circle | radius: {self.radius} | area: {self.area():.2f}")

# Derived Class - Rectangle:
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = float(length)
        self.breadth = float(breadth)

    def area(self):
        return self.length * self.breadth

    def display(self):
        print(f"Shape: Rectangle | length: {self.length}, breadth: {self.breadth} | area: {self.area():.2f}")
```

### Area calculation for Circle and Rectangle: 

```python
circle = Circle(radius=2.4)
circle.display()
```

```python
rectangle = Rectangle(length=5.0, breadth=3.0)
rectangle.display()
```
