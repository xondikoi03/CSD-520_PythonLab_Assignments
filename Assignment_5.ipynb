## Problem 1:

Write a program that reads `words.txt` and prints only the words with more than 20 characters.

````python
# Reading a text file:
read_file = open("resources/text_random.txt", "r")

#  Print only the words with more than 20 charcters:
for line in read_file:
    words = line.split() # Splits the line into list of words
    for word in words:
        if len(word) > 20:
            print(word)

read_file.close() # Close the file after reading
````

## Problem 2:

Write a function in python to count the number of lines from a text file `story.txt` which is not starting with an alphavet "T".

```
Example: If the file "story.txt"

A boy playing there.
There is a playground.
An aeroplane is in the sky.
The sky is pink.
Alphabets and numbers are allowed in the password.
```

````python
# Read the file:
story_file = open("resources/story.txt", "r")

# Function to count the number of lines that don't start with the letter "T":
def count_lines(file):
    line_count = 0
    for line in file:
        if line[0] != "T":
            line_count += 1
    return line_count

# Count the number of lines that don't start with the letter "T" in the story file:
num_lines = count_lines(story_file)
# Close the file:
story_file.close()
print(f"The number of lines that don't start with the letter 'T' in the story file is: {num_lines}")
````

## Problem 3:

Write a program to accept two files from the user. Copy the text from first file and copy it into another file. 

````python
# Accepting user input file:
user_file = input("Enter the name of the file to read: ")
write_file = open(user_file, "w")

# Write some strings to the file:
write_file.write("My name is Raj Handique.\n")
write_file.write("CSD-520 Python Programming.\n")
write_file.write("This is assignment is on file handling.\n")
write_file.close()

# Read the file:
read_file = open(user_file, "r")
text = read_file.read()
read_file.close()

# Accepting user input file to copy the text:
user_file1 = input("Enter the name of file to copy: ")
# Copy the text from the user input file to a new file:
copy_file = open(user_file1, "w")
copy_file.write(text)

# Close the file after copying:
copy_file.close()    
````
